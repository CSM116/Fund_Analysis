```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import colors
from datetime import date

import plotly.graph_objects as go
from plotly.subplots import make_subplots
import plotly.io as pio
pio.renderers.default = 'notebook'

import ptm_lib as ptm
import DoR_lib as dor

%load_ext autoreload
%autoreload 2
```

```python
from_date = date(1900, 1, 1) 
to_date = date.today()

file_name = "Int_Comrc"
tickers = ['OCDDY', 'CARG', 'JTKWY', 'GLBE', 'ETSY', 'W', 
           'ATDRY', 'CART', 'CHWY', 'EXPE', 'YAHOY', 'EBAY', 
           'JD', 'CPNG', 'CVNA', 'MELI', 'PDD', 'BKNG', 'BABA', 
           'PROSY', 'AMZN'
          ]

# Frequency mapping: yfinance supports specific interval strings
freq_map = {
    'Daily': '1d'
}

# Output file
output_file = f'Tickers/{file_name}.xlsx'

# Write to Excel
with pd.ExcelWriter(output_file, engine='xlsxwriter') as writer:
    for label, yf_freq in freq_map.items():
        df = dor.download_and_clean(tickers, from_date, to_date, freq=yf_freq, singleStock=False)
        df.to_excel(writer, sheet_name=label)

print(f"Data saved to {output_file}")
```
